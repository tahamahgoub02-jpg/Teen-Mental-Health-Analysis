```python
# Import necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt



# Set a modern style
sns.set_theme(style="whitegrid")

# Create an advanced boxplot
plt.figure(figsize=(12, 7))
sns.boxplot(
    data=df,
    x="platform_usage",          # Grouping by social media platform
    y="daily_social_media_hours",# Distribution of daily usage
    hue="gender",                # Separate by gender
    palette="Set2",              # Color palette
    dodge=True                   # Separate boxes for male/female
)

# Add swarmplot for individual data points
sns.swarmplot(
    data=df,
    x="platform_usage",
    y="daily_social_media_hours",
    hue="gender",
    dodge=True,
    palette="dark:.3",
    alpha=0.6
)

# Titles and labels
plt.title("Daily Social Media Hours by Platform and Gender", fontsize=16, weight="bold")
plt.xlabel("Platform Usage", fontsize=12)
plt.ylabel("Daily Social Media Hours", fontsize=12)

# Adjust legend
plt.legend(title="Gender", bbox_to_anchor=(1.05, 1), loc='upper left')

plt.show()
```

```python
males_df = df[df['gender'] == 'male'].copy()
print(f"Total male records: {len(males_df)}")
print(f"Academic performance range: {males_df['academic_performance'].min():.2f} - {males_df['academic_performance'].max():.2f}")
```

```python
# Cell 4 - Bin Academic Performance into Categories
bins = [0, 2.5, 3.0, 3.5, 4.0]
labels = ['Low (2.0–2.5)', 'Below Avg (2.5–3.0)', 'Above Avg (3.0–3.5)', 'High (3.5–4.0)']

males_df['perf_category'] = pd.cut(
    males_df['academic_performance'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

perf_counts = males_df['perf_category'].value_counts().sort_index()
print(perf_counts)
```

```python

# Filter: Males with addiction_level >= 8
males_high_addiction = df[(df['gender'] == 'male') & (df['addiction_level'] >= 8)].copy()
print(f"Total males with addiction level 8+: {len(males_high_addiction)}")

# Bin Academic Performance into Categories
bins = [0, 2.5, 3.0, 3.5, 4.0]
labels = ['Low (2.0–2.5)', 'Below Avg (2.5–3.0)', 'Above Avg (3.0–3.5)', 'High (3.5–4.0)']

males_high_addiction['perf_category'] = pd.cut(
    males_high_addiction['academic_performance'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

perf_counts = males_high_addiction['perf_category'].value_counts().sort_index()
print(perf_counts)

# Plot Pie Chart
colors = ['#e74c3c', '#e67e22', '#3498db', '#2ecc71']
explode = [0.04] * len(perf_counts)

fig, ax = plt.subplots(figsize=(8, 8))

wedges, texts, autotexts = ax.pie(
    perf_counts,
    labels=perf_counts.index,
    autopct='%1.1f%%',
    colors=colors,
    explode=explode,
    startangle=140,
    pctdistance=0.78,
    textprops={'fontsize': 11}
)

for autotext in autotexts:
    autotext.set_fontweight('bold')
    autotext.set_fontsize(11)

ax.set_title(
    'Academic Performance of Males with High Addiction Level (≥ 8)',
    fontsize=14, fontweight='bold', pad=20
)

legend_labels = [f"{lbl}  (n={cnt})" for lbl, cnt in zip(perf_counts.index, perf_counts.values)]
ax.legend(
    wedges, legend_labels,
    title='Performance Category',
    loc='lower center',
    bbox_to_anchor=(0.5, -0.12),
    ncol=2,
    fontsize=10
)

plt.tight_layout()
plt.savefig('male_high_addiction_academic_pie.png', dpi=150, bbox_inches='tight')
plt.show()

# Summary
print("\n=== Summary Statistics ===")
pct = (perf_counts / perf_counts.sum() * 100).round(1)
summary = pd.DataFrame({'Count': perf_counts, 'Percentage': pct})
print(summary)
```

```python

# Filter: Females only
females_df = df[df['gender'] == 'female'].copy()
print(f"Total female records: {len(females_df)}")

# Plot Curve Graph
plt.figure(figsize=(10, 6))

# KDE / Curve plot using seaborn
sns.kdeplot(
    data=females_df,
    x='stress_level',
    hue='addiction_level',
    palette='coolwarm',
    linewidth=2,
    legend=True
)

plt.title(
    'Relationship Between Stress Level and Addiction Level for Females',
    fontsize=14, fontweight='bold', pad=15
)
plt.xlabel('Stress Level', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('female_stress_vs_addiction_curve.png', dpi=150, bbox_inches='tight')
plt.show()

# Summary Statistics
print("\n=== Correlation: Stress Level vs Addiction Level (Females) ===")
correlation = females_df['stress_level'].corr(females_df['addiction_level'])
print(f"Pearson Correlation Coefficient: {correlation:.4f}")

print("\n=== Mean Addiction Level by Stress Level ===")
summary = females_df.groupby('stress_level')['addiction_level'].mean().round(2)
print(summary)
```

```python
# Import libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt



filtered = df[df["platform_usage"] != "Both"]

# Count depression cases per platform
depression_counts = (
    filtered[filtered["depression_label"] == 1]
    .groupby("platform_usage")["depression_label"]
    .count()
    .sort_values(ascending=False)
)

# Convert to DataFrame for plotting
depression_df = depression_counts.reset_index()
depression_df.columns = ["Platform", "Depression Cases"]

# Create a smooth curve plot
plt.figure(figsize=(10,6))
sns.lineplot(
    data=depression_df,
    x="Platform",
    y="Depression Cases",
    marker="o",
    linewidth=2.5,
    color="crimson"
)

# Add titles and labels
plt.title("Curve of Depression Cases by Platform ", fontsize=16, weight="bold")
plt.xlabel("Platform", fontsize=12)
plt.ylabel("Number of Depression Cases", fontsize=12)

# Annotate points
for i, row in depression_df.iterrows():
    plt.text(i, row["Depression Cases"] + 0.2, row["Depression Cases"], 
             ha="center", fontsize=12, color="black")

plt.show()
```
